# Task 1
##### The factory makes suits and coats.
## Given Data

| Product | Fabric per unit (m) | Labor per unit (hours) | Profit per unit (rubles) |
|--------|----------------------|-------------------------|---------------------------|
| Suit   | 3                    | 4                       | 5000                     |
| Coat   | 2                    | 2                       | 3000                     |

### Decision Variables

Let:

- $x_1$ = number of suits produced
- $x_2$ = number of coats produced

### Objective Function

The goal is to maximize the total profit.

Each suit gives a profit of 5000 rubles, and each coat gives a profit of 3000 rubles.

Therefore, the objective function is:

$$
\text{Maximize } Z = 5000x_1 + 3000x_2
$$

### Constraints

#### 1. Fabric constraint

Each suit requires 3 meters of fabric and each coat requires 2 meters.

The factory has 180 meters of fabric:

$$
3x_1 + 2x_2 \leq 180
$$

#### 2. Labor constraint

Each suit requires 4 hours of labor and each coat requires 2 hours.

The factory has 200 hours of labor:

$$
4x_1 + 2x_2 \leq 200
$$

#### 3. Non-negativity constraints

$$
x_1 \geq 0
$$

$$
x_2 \geq 0
$$

```python
import pyomo.environ as pyo

# Create the model
model = pyo.ConcreteModel()

# Decision variables
model.x1 = pyo.Var(domain=pyo.NonNegativeReals)
model.x2 = pyo.Var(domain=pyo.NonNegativeReals)

# Objective function
model.profit = pyo.Objective(
    expr=5000 * model.x1 + 3000 * model.x2,
    sense=pyo.maximize
)

# Fabric constraint
model.fabric = pyo.Constraint(
    expr=3 * model.x1 + 2 * model.x2 <= 180
)

# Labor constraint
model.labor = pyo.Constraint(
    expr=4 * model.x1 + 2 * model.x2 <= 200
)

# Solver
solver = pyo.SolverFactory('appsi_highs')

# Solve
results = solver.solve(model)

# Results
print("Number of suits:", pyo.value(model.x1))
print("Number of coats:", pyo.value(model.x2))
print("Maximum profit:", pyo.value(model.profit))
```

# Task 2. Optimization of the Daily Diet
## Given Data

| Product | Cost (rubles/kg) | Vitamin A (units/kg) | Vitamin C (units/kg) |
| ------- | ---------------: | -------------------: | -------------------: |
| P1      |              120 |                  200 |                  100 |
| P2      |               80 |                  100 |                  200 |

### Daily Requirements

* Vitamin A: at least 600 units
* Vitamin C: at least 500 units

## Decision Variables

Let:

* $x_1$ = amount of Product P1 in the daily diet, in kilograms
* $x_2$ = amount of Product P2 in the daily diet, in kilograms

## Objective Function

The goal is to minimize the total cost of the daily diet.

Product P1 costs 120 rubles per kg, so its cost is:

$$
120x_1
$$

Product P2 costs 80 rubles per kg, so its cost is:

$$
80x_2
$$

Therefore, the total cost is:

$$
Z = 120x_1 + 80x_2
$$

We want to minimize this value:

$$
\text{Minimize } Z = 120x_1 + 80x_2
$$


## Constraints

### 1. Vitamin A constraint

Product P1 provides 200 units of Vitamin A per kg, while Product P2 provides 100 units per kg.

The diet must provide **at least 600 units** of Vitamin A.

Therefore:

$$
200x_1 + 100x_2 \geq 600
$$

### 2. Vitamin C constraint

Product P1 provides 100 units of Vitamin C per kg, while Product P2 provides 200 units per kg.

The diet must provide **at least 500 units** of Vitamin C.

Therefore:

$$
100x_1 + 200x_2 \geq 500
$$

### 3. Non-negativity constraints

The amount of each product cannot be negative:

$$
x_1 \geq 0
$$

$$
x_2 \geq 0
$$


```python
model = pyo.ConcreteModel()

model.x1 = pyo.Var(domain=pyo.NonNegativeReals)
model.x2 = pyo.Var(domain=pyo.NonNegativeReals)

model.cost = pyo.Objective(
    expr = 120 * model.x1 + 80 * model.x2,
    sense = pyo.minimize
)

model.VitA = pyo.Constraint(
    expr = 200 * model.x1 + 100 * model.x2 >= 600
)
model.VitC = pyo.Constraint(
    expr = 100 * model.x1 + 200 * model.x2 >= 500
)

solver = pyo.SolverFactory('appsi_highs')
result = solver.solve(model)

print("Number of suits:", pyo.value(model.x1))
print("Number of coats:", pyo.value(model.x2))
print("Maximum profit:", pyo.value(model.cost))
```

# Task 3. Investment Portfolio Optimization

## Problem Statement

An investor has 1,000,000 rubles and plans to invest this money in two projects: Project A and Project B.

Project A provides an annual return of 12%.

Project B provides an annual return of 18%.

Because of risk restrictions:
- The investment in Project B cannot exceed 400,000 rubles.
- The investment in Project A must be at least 30% of the total investment amount.

The goal is to determine how much money should be invested in each project in order to maximize the total annual income.

## Given Data

| Project | Available Investment | Annual Return |
|---|---:|---:|
| Project A | At least 30% of total investment | 12% |
| Project B | At most 400,000 rubles | 18% |

### Total available capital

- 1,000,000 rubles

## Mathematical Model

Let:

- $x_A$ = amount invested in Project A, in rubles
- $x_B$ = amount invested in Project B, in rubles

We want to maximize the total annual income:

$$
\text{Maximize } Z = 0.12x_A + 0.18x_B
$$

subject to:

$$
x_A + x_B \leq 1000000
$$

$$
x_B \leq 400000
$$

$$
x_A \geq 300000
$$

$$
x_A, x_B \geq 0
$$

```python
model = pyo.ConcreteModel()

model.xA = pyo.Var(domain=pyo.NonNegativeReals)
model.xB = pyo.Var(domain=pyo.NonNegativeReals)

model.income = pyo.Objective(
    expr = 0.12 * model.xA + 0.18 * model.xB,
    sense = pyo.maximize
)

model.total = pyo.Constraint(
    expr=model.xA + model.xB <= 1000000
)

model.B_limit = pyo.Constraint(
    expr=model.xB <= 400000
)

model.A_min = pyo.Constraint(
    expr=model.xA >= 300000
)

solver = pyo.SolverFactory('appsi_highs')
result = solver.solve(model)

print("Investment in Project A:", pyo.value(model.xA))
print("Investment in Project B:", pyo.value(model.xB))
print("Maximum annual income:", pyo.value(model.income))
```

# Task 4. Motor Fuel Blending

A refinery produces "Standard" gasoline by mixing three components: K1, K2, and K3.

| Component | Cost (USD/barrel) | Octane Number |
|---|---:|---:|
| K1 | 30 | 85 |
| K2 | 40 | 92 |
| K3 | 50 | 98 |

The finished gasoline must have:
- an octane number of at least 93;
- a total volume of at least 10,000 barrels;
- K1 must not exceed 20% of the total mixture.

The goal is to minimize the total cost of the mixture.

## Mathematical Model

Let:

- $x_1$ = barrels of K1
- $x_2$ = barrels of K2
- $x_3$ = barrels of K3

Minimize:

$$
Z = 30x_1 + 40x_2 + 50x_3
$$

subject to the volume requirement:

$$
x_1 + x_2 + x_3 \geq 10000
$$

The average octane number must be at least 93:

$$
85x_1 + 92x_2 + 98x_3 \geq 93(x_1+x_2+x_3)
$$

K1 can be at most 20% of the total mixture:

$$
x_1 \leq 0.20(x_1+x_2+x_3)
$$

and:

$$
x_1,x_2,x_3 \geq 0
$$

```python
model = pyo.ConcreteModel()

model.x1 = pyo.Var(domain=pyo.NonNegativeReals)
model.x2 = pyo.Var(domain=pyo.NonNegativeReals)
model.x3 = pyo.Var(domain=pyo.NonNegativeReals)

model.cost = pyo.Objective(
    expr=30*model.x1 + 40*model.x2 + 50*model.x3,
    sense=pyo.minimize
)

model.volume = pyo.Constraint(
    expr=model.x1 + model.x2 + model.x3 >= 10000
)

model.octane = pyo.Constraint(
    expr=85*model.x1 + 92*model.x2 + 98*model.x3
         >= 93*(model.x1 + model.x2 + model.x3)
)

model.K1_limit = pyo.Constraint(
    expr=model.x1 <= 0.20*(model.x1 + model.x2 + model.x3)
)

solver = pyo.SolverFactory('appsi_highs')
result = solver.solve(model)

print("K1:", pyo.value(model.x1), "barrels")
print("K2:", pyo.value(model.x2), "barrels")
print("K3:", pyo.value(model.x3), "barrels")
print("Minimum cost:", pyo.value(model.cost), "USD")
```

# Task 5. Transportation Problem

Two warehouses supply three stores.

### Supply

- Warehouse C1: 150 units
- Warehouse C2: 200 units

### Demand

- Store M1: 100 units
- Store M2: 120 units
- Store M3: 130 units

### Transportation Cost per Unit

| | M1 | M2 | M3 |
|---|---:|---:|---:|
| C1 | 8 | 6 | 10 |
| C2 | 9 | 5 | 7 |

The goal is to minimize the total transportation cost.

## Mathematical Model

Let $x_{ij}$ represent the number of units transported from warehouse $C_i$ to store $M_j$.

The objective is:

$$
\text{Minimize } Z =
8x_{11}+6x_{12}+10x_{13}
+9x_{21}+5x_{22}+7x_{23}
$$

Warehouse supply constraints:

$$
x_{11}+x_{12}+x_{13} \leq 150
$$

$$
x_{21}+x_{22}+x_{23} \leq 200
$$

Store demand constraints:

$$
x_{11}+x_{21}=100
$$

$$
x_{12}+x_{22}=120
$$

$$
x_{13}+x_{23}=130
$$

All transportation amounts must be non-negative.

```python
model = pyo.ConcreteModel()

# Decision variables
model.x11 = pyo.Var(domain=pyo.NonNegativeReals)
model.x12 = pyo.Var(domain=pyo.NonNegativeReals)
model.x13 = pyo.Var(domain=pyo.NonNegativeReals)
model.x21 = pyo.Var(domain=pyo.NonNegativeReals)
model.x22 = pyo.Var(domain=pyo.NonNegativeReals)
model.x23 = pyo.Var(domain=pyo.NonNegativeReals)

# Objective
model.cost = pyo.Objective(
    expr=8*model.x11 + 6*model.x12 + 10*model.x13
       + 9*model.x21 + 5*model.x22 + 7*model.x23,
    sense=pyo.minimize
)

# Warehouse supply
model.C1_supply = pyo.Constraint(
    expr=model.x11 + model.x12 + model.x13 <= 150
)

model.C2_supply = pyo.Constraint(
    expr=model.x21 + model.x22 + model.x23 <= 200
)

# Store demand
model.M1_demand = pyo.Constraint(
    expr=model.x11 + model.x21 == 100
)

model.M2_demand = pyo.Constraint(
    expr=model.x12 + model.x22 == 120
)

model.M3_demand = pyo.Constraint(
    expr=model.x13 + model.x23 == 130
)

# Solve
solver = pyo.SolverFactory('appsi_highs')
result = solver.solve(model)

print("C1 -> M1:", pyo.value(model.x11))
print("C1 -> M2:", pyo.value(model.x12))
print("C1 -> M3:", pyo.value(model.x13))
print("C2 -> M1:", pyo.value(model.x21))
print("C2 -> M2:", pyo.value(model.x22))
print("C2 -> M3:", pyo.value(model.x23))
print("Minimum transportation cost:", pyo.value(model.cost))
```

# Task 6. Plywood Cutting Optimization

A furniture factory needs:

- 120 pieces of type A
- 150 pieces of type B

Three cutting patterns are available:

| Pattern | A pieces | B pieces |
|---|---:|---:|
| 1 | 3 | 1 |
| 2 | 1 | 4 |
| 3 | 2 | 2 |

The goal is to determine how many sheets should be cut using each pattern while minimizing the total number of sheets.

## Mathematical Model

Let:

- $x_1$ = number of sheets cut using pattern 1
- $x_2$ = number of sheets cut using pattern 2
- $x_3$ = number of sheets cut using pattern 3

Minimize:

$$
Z=x_1+x_2+x_3
$$

subject to the required pieces:

$$
3x_1+x_2+2x_3 \geq 120
$$

$$
x_1+4x_2+2x_3 \geq 150
$$

$$
x_1,x_2,x_3 \geq 0
$$

```python
model = pyo.ConcreteModel()

model.x1 = pyo.Var(domain=pyo.NonNegativeReals)
model.x2 = pyo.Var(domain=pyo.NonNegativeReals)
model.x3 = pyo.Var(domain=pyo.NonNegativeReals)

# Minimize total sheets
model.sheets = pyo.Objective(
    expr=model.x1 + model.x2 + model.x3,
    sense=pyo.minimize
)

# Required A pieces
model.A = pyo.Constraint(
    expr=3*model.x1 + model.x2 + 2*model.x3 >= 120
)

# Required B pieces
model.B = pyo.Constraint(
    expr=model.x1 + 4*model.x2 + 2*model.x3 >= 150
)

solver = pyo.SolverFactory('appsi_highs')
result = solver.solve(model)

print("Pattern 1:", pyo.value(model.x1))
print("Pattern 2:", pyo.value(model.x2))
print("Pattern 3:", pyo.value(model.x3))
print("Minimum number of sheets:", pyo.value(model.sheets))
```

# Task 7. Staff Scheduling

A call center needs enough employees to cover three 4-hour periods:

| Time | Required Employees |
|---|---:|
| 08:00–12:00 | 15 |
| 12:00–16:00 | 25 |
| 16:00–20:00 | 20 |

Employees work 8-hour shifts:

- Shift 1: 08:00–16:00, costing 1600 rubles per shift
- Shift 2: 12:00–20:00, costing 1800 rubles per shift

The goal is to minimize the total payroll while satisfying the staffing requirements.

## Mathematical Model

Let:

- $x_1$ = number of employees working Shift 1
- $x_2$ = number of employees working Shift 2

Minimize:

$$
Z=1600x_1+1800x_2
$$

The 08:00–12:00 period is covered only by Shift 1:

$$
x_1\geq15
$$

The 12:00–16:00 period is covered by both shifts:

$$
x_1+x_2\geq25
$$

The 16:00–20:00 period is covered only by Shift 2:

$$
x_2\geq20
$$

$$
x_1,x_2\geq0
$$

```python
model = pyo.ConcreteModel()

# Number of employees in each shift
model.x1 = pyo.Var(domain=pyo.NonNegativeIntegers)
model.x2 = pyo.Var(domain=pyo.NonNegativeIntegers)

# Minimize payroll
model.payroll = pyo.Objective(
    expr=1600*model.x1 + 1800*model.x2,
    sense=pyo.minimize
)

# Staffing constraints
model.morning = pyo.Constraint(
    expr=model.x1 >= 15
)

model.afternoon = pyo.Constraint(
    expr=model.x1 + model.x2 >= 25
)

model.evening = pyo.Constraint(
    expr=model.x2 >= 20
)

solver = pyo.SolverFactory('appsi_highs')
result = solver.solve(model)

print("Shift 1 employees:", pyo.value(model.x1))
print("Shift 2 employees:", pyo.value(model.x2))
print("Minimum payroll:", pyo.value(model.payroll))
```

# Task 9. Inventory Planning

The company plans production for three months.

| Month | Demand | Production Cost per Unit |
|---|---:|---:|
| 1 | 100 | 50 rubles |
| 2 | 150 | 55 rubles |
| 3 | 200 | 60 rubles |

Additional information:

- Initial inventory = 20 units
- Storage cost = 4 rubles per unit carried from one month to the next
- Maximum production = 180 units per month

The goal is to minimize total production and storage costs.

## Mathematical Model

Let:

- $x_t$ = production in month $t$
- $s_t$ = inventory remaining at the end of month $t$

The objective is:

$$
\text{Minimize } Z =
50x_1+55x_2+60x_3
+4s_1+4s_2+4s_3
$$

Inventory balance:

$$
s_1=20+x_1-100
$$

$$
s_2=s_1+x_2-150
$$

$$
s_3=s_2+x_3-200
$$

Production capacity:

$$
x_t\leq180
$$

for each month.

Inventory and production must be non-negative:

$$
x_t\geq0,\quad s_t\geq0
$$

```python
model = pyo.ConcreteModel()

# Production variables
model.x1 = pyo.Var(domain=pyo.NonNegativeReals)
model.x2 = pyo.Var(domain=pyo.NonNegativeReals)
model.x3 = pyo.Var(domain=pyo.NonNegativeReals)

# Inventory variables
model.s1 = pyo.Var(domain=pyo.NonNegativeReals)
model.s2 = pyo.Var(domain=pyo.NonNegativeReals)
model.s3 = pyo.Var(domain=pyo.NonNegativeReals)

# Minimize production + storage costs
model.cost = pyo.Objective(
    expr=50*model.x1 + 55*model.x2 + 60*model.x3
       + 4*model.s1 + 4*model.s2 + 4*model.s3,
    sense=pyo.minimize
)

# Inventory balance
model.balance1 = pyo.Constraint(
    expr=model.s1 == 20 + model.x1 - 100
)

model.balance2 = pyo.Constraint(
    expr=model.s2 == model.s1 + model.x2 - 150
)

model.balance3 = pyo.Constraint(
    expr=model.s3 == model.s2 + model.x3 - 200
)

# Production capacity
model.capacity1 = pyo.Constraint(expr=model.x1 <= 180)
model.capacity2 = pyo.Constraint(expr=model.x2 <= 180)
model.capacity3 = pyo.Constraint(expr=model.x3 <= 180)

solver = pyo.SolverFactory('appsi_highs')
result = solver.solve(model)

print("Month 1 production:", pyo.value(model.x1))
print("Month 2 production:", pyo.value(model.x2))
print("Month 3 production:", pyo.value(model.x3))

print("Month 1 inventory:", pyo.value(model.s1))
print("Month 2 inventory:", pyo.value(model.s2))
print("Month 3 inventory:", pyo.value(model.s3))

print("Minimum total cost:", pyo.value(model.cost))
```

# Task 10. Power Generation Optimization

A power plant has two generators:

| Generator | Maximum Power | Cost |
|---|---:|---:|
| G1 | 100 MW | 1500 rubles/MWh |
| G2 | 150 MW | 1200 rubles/MWh |

The load schedule is:

| Period | Duration | Demand |
|---|---:|---:|
| Day | 12 hours | 180 MW |
| Night | 12 hours | 80 MW |

Due to technological restrictions, G1 must generate at least 20 MW during both periods.

The goal is to determine the generation level of each generator in each period while minimizing total daily costs.

## Mathematical Model

Let:

- $P_{1D}$ = power generated by G1 during the day
- $P_{2D}$ = power generated by G2 during the day
- $P_{1N}$ = power generated by G1 at night
- $P_{2N}$ = power generated by G2 at night

The total daily cost is:

$$
Z =
12(1500P_{1D}+1200P_{2D})
+
12(1500P_{1N}+1200P_{2N})
$$

Minimize $Z$ subject to the demand requirements:

$$
P_{1D}+P_{2D}=180
$$

$$
P_{1N}+P_{2N}=80
$$

Generator limits:

$$
20\leq P_{1D}\leq100
$$

$$
20\leq P_{1N}\leq100
$$

$$
0\leq P_{2D}\leq150
$$

$$
0\leq P_{2N}\leq150
$$

```python
model = pyo.ConcreteModel()

# Generation variables
model.P1D = pyo.Var(bounds=(20, 100))
model.P2D = pyo.Var(bounds=(0, 150))
model.P1N = pyo.Var(bounds=(20, 100))
model.P2N = pyo.Var(bounds=(0, 150))

# Minimize daily generation cost
model.cost = pyo.Objective(
    expr=12*(1500*model.P1D + 1200*model.P2D)
       + 12*(1500*model.P1N + 1200*model.P2N),
    sense=pyo.minimize
)

# Demand constraints
model.day_demand = pyo.Constraint(
    expr=model.P1D + model.P2D == 180
)

model.night_demand = pyo.Constraint(
    expr=model.P1N + model.P2N == 80
)

# Solve
solver = pyo.SolverFactory('appsi_highs')
result = solver.solve(model)

print("G1 daytime:", pyo.value(model.P1D), "MW")
print("G2 daytime:", pyo.value(model.P2D), "MW")
print("G1 nighttime:", pyo.value(model.P1N), "MW")
print("G2 nighttime:", pyo.value(model.P2N), "MW")
print("Minimum daily cost:", pyo.value(model.cost), "rubles")
```
